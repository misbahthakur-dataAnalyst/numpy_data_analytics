```python
import numpy as np
import statistics as stats
baked_food = [200,300,150,130,280,200,170,188]
a= np.array(baked_food)
print(np.mean(baked_food)) #sum of all the values divided by number of values
print(np.median(baked_food)) #center value after sorting
print(stats.mode(baked_food)) #mode is most reoccured 
print(np.std(baked_food)) #standard divation how far values are (underroot value daita hai) spread 
print(np.var(baked_food)) #variance (underroot ka square hota hai)

```

```python
# -1 represents inversely proportional relationship
# 1 represents proportional relationship
# 0 means no relationship
tobacco_consumption= [30,50,10,30,40,50]
deaths= [100,120,70,100,120,112]


print(np.corrcoef([tobacco_consumption,deaths]))
```

```python
price = [300,100,350,150,200]
sales = [10,4,20,7,17,3]

print(np.corrcoef([price,sales])) #negative relationship
```
