```python
# inspecting an array

import numpy as np

a= [[30,40,40],[20,40,30.8]]
arr= np.array(a)
print(arr)
print(arr.shape) #rows, colums
print(len(arr)) #number  of nested values
print(np.size(arr)) #number of elements
print(type(arr)) #type of variable
print(arr.dtype) #data type of array
print(arr.astype(int)) #conversion of data type
```
